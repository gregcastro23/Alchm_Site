# RecipeGenerator Component Documentation

This document describes the `RecipeGenerator` component, its functionality, and its integration within the Alchm Site application.

## Introduction

The `RecipeGenerator` component is a React component designed to provide users with personalized food recommendations based on their astrological birth data. It aims to offer a unique and engaging feature within the Alchm Site application.

## Component Overview

-   **Location:** `src/components/recipegenerator/RecipeGenerator.js`
-   **Purpose:** To generate food recommendations based on astrological birth data.
-   **Key Features:**
    -   Input fields for birth date and time.
    -   API integration to fetch astrological chart and recommendations.
    -   Display of astrological chart data.
    -   Display of food recommendations.
    -   Loading and error state handling.

## How to Use

Navigate to the `/recipe-generator` route in the application to access the component. Enter your birth date and time, and the component will display your astrological data and personalized recipe recommendations.

## Technical Details

-   **Framework/Library:** React.js
-   **Styling:** Uses `recipegenerator.css` (located at `src/components/recipegenerator/recipegenerator.css`).
-   **API Interaction:**
    -   Uses `axios` to make a POST request to the backend API endpoint `/api/recipe-generator`.
    -   **Request Payload:**
        ```json
        {
          "birthDate": "YYYY-MM-DD",
          "birthTime": "HH:MM:SS"
        }
        ```
    -   **Response Structure:**
        ```json
        {
          "chart": { ... astrological data object ... },
          "recommendations": ["Recipe 1", "Recipe 2", ...]
        }
        ```
-   **State Management:** Uses `useState` for managing form inputs, loading state, errors, chart data, and recommendations.

## Dependencies

-   React.js
-   `axios` (for API requests)
-   `recipegenerator.css` (for styling)
